```python
for i in range(1, 101):
    if i % 3 == 0:
        print("F")
    elif i % 5 == 0:
        print("B")
    else :
        print(i)
```

```python

for i in range(1,20):
    print("F"*(i % 3 ==0)+"B"*(i%5==0) or i)
```

```python
# 20.7 연습문제
for i in range(1, 101):
    if i % 2 == 0 and i % 11 == 0 :
        print("FizzBuzz")
    elif i % 2 == 0 :
        print("Fizz")
    elif i % 11 == 0 :
        print("FizzBuzz")
    else :
        print(i)
```

```python
# 20.8 심사문제
a, b = map(int,input().split())
for i in range(a,b+1):
    print("Fizz"*(i%5==0)+"Buzz"*(i%7==0) or i)
```
