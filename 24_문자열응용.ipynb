```python
# 문자열 서식 지정자와 포매팅 사용하기

print("I am %s" % "James")

name = "Kelly"
print("I am %s" % name)

age = 20
rou = 3.14
print("I am %d years old" % age)
print("I am %f years old" % rou)

print("%10s" % "Python")
print("%-10s" % "Python")

print("Today is %d %s" % (3, "April"))
```
