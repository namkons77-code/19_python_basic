```python
# 위치인수
def print_numbers(a, b, c):
    print(a)
    print(b)
    print(c)

print_numbers(10, 20, 30)
```

```python
# 언패킹 사용하기
x = [10, 20, 30]
print_numbers(*x)
```

```python
# # 가변인수 함수 만들기
# def 함수이름(*매개변수):
#     코드

def print_numbers(*args):
    print(args)
    for arg in args:
        print(arg)

print_numbers(10)
print_numbers(10, 20)

```

```python
# 키워드 인수
def persoal_info(name, age, address):
    print("이름:", name)
    print("나이:", age)
    print("주소:", address)

persoal_info("길동", 30, "서울")

persoal_info(age=30, address="대구", name="닐동")

```

```python
x = {
    "name":"길동", "age":30, "address":"용산"
}
persoal_info(*x)
persoal_info(**x)
```

```python
# # 가변 키워드 인수 함수 만들기
# def 함수이름(**매개변수):
#     코드

def persoal_info(**kwargs):
    print(kwargs)
    for kw, arg in kwargs.items():
        print(kw, arg)

persoal_info(name="딜동", age=30)
```

```python
# 매개변수의 초깃값 지정하기
def persoal_info(name, age, address="비공개", sex="비"):  # 디폴트는 마지막으로...
    print("이름:", name)
    print("나이:", age)
    print("주소:", address)
    print(sex)

persoal_info("릴동", 30)
persoal_info("릴동", 30, "용산")
```
