```python
# while 조건식:
#   반복할 코드
#   변화식
```

```python
i = 0
while i < 10:
    print("hello", i)
    i += 1
```

```python
i = 10
while i > 0 :
    print(i)
    i-=1
```

```python
x = int(input())
i = 0
while i < x :
    print(i)
    i += 1
```

```python
import random
i = 0
while i != 2:
    i = random.randint(1,10)
    print(i)
```

```python
# 17.5 연습문제
i = 2
j = 5
while i <= 32 or j >= 1:
    print(i,j)
    i *= 2
    j -= 1

```

```python
# 17.6 심사문제
x = int(input())
while  x >= 1350 :
    x -= 1350
    print(x)
```
