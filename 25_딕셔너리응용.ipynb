```python
x = {
    "a":10, "b":20, "c":30, "d":40
}

x.setdefault("f",100)
print(x)
x.update(a=90)
print(x)
x.pop("a")
print(x)
x.get("c")
print(x)
x.get("z")
print(x)

print(x.items())
print(x.keys())
print(sum(x.values()))
```

```python
# 반복문으로 딕셔너리의 키-값 가져오기
x = {
    "a":10, "b":20, "c":30, "d":40
}

for i in x:
    print(i)
for i in x.values():
    print(i)
for i in x.items():
    print(i)
```
