```python
# 리스트에 요소 하나 추가하기

a = [10, 20, 30]
a.append(40)
a.append([50, 60])
a.extend([70, 80])
print(a)
len(a)
```

```python
a = [10, 20, 30]

a.insert(2, 500)
print(a)

a.pop(1)
print(a)

a.remove(30)
print(a)
```

```python
a = [10, 20, 30, 15, 20, 40]

print(a.index(20))
print(a.count(20))

a.reverse()
print(a)

a.sort()
print(a)

a.sort(reverse=True)
print(a)


```

```python
# 리스트의 할당과 복사
a = [0, 0, 0, 0, 0, 0]
b = a

b[2] = 99
print(b)

c = a.copy()
print(a, b, c)
```

```python
# 반복문으로 리스트 요소 출력

d = [38, 21, 53, 62, 19]
for i in d:
    print(i)

for x, y in enumerate(d):
    print(x, y)
```

```python
# while
a = [38, 21, 53, 62, 19]

i = 0
while i < len(a):
    print(a[i])
    i+=1
```

```python
a = [38, 21, 53, 62, 19]
x = 0
for i in a:
    x += i
print(x)

sum(a)
min(a)
max(a)
```

```python
# 리스트 표현식 사용하기
# [식 for 변수 in 리스트
a = [i for i in range(10)]
print(a)

a=[]
for i in range(10):
    a.append(i)
print(a)

a = [i+5 for i in range(10)]
print(a)

a = [i+5 for i in range(10) if i%2==1]
print(a)

```

```python
a = [ i * j for i in range(1,10) for j in range(1,10) ]
print(a)
```

```python
# # 리스트에 map 사용하기
# list(map(함수, 리스트))
# tuple(map(함수, 튜플))

a = [1.2, 2.5, 3.7, 4.6]
b = list(map(int, a))
print(b)
```

```python
# 튜플 응용
a = (38, 21, 53, 62, 19, 53)
a.index(19)

b = tuple(i for i in range(10) if i%2==0)
print(b)
```

```python
# 22.9 연습문제
a = ['alpha', 'bravo', 'charlie', 'delta', 'echo', 'foxtrot', 'golf', 'hotel', 'india']
b = [i for i in a if len(i)==5]
print(b)
```

```python
# 22.10 심사문제
a, b = map(int, input().split())
x = [ 2**i for i in range(a,b+1) ]
x.pop(1)
x.pop(-2)
print(x)
```
