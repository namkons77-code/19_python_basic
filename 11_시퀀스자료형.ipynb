```python
# 시퀀스 자료형(list, tuple, range, 문자열)의 공통 기능 사용

a = [10, 20, 30, 40, 50, 60, 70, 80, 90]

print(a[0])  # 첫 번째 요소
print(a[-1])  # 마지막 요소 
print(a[1:4])  # 두 번째 요소부터 네 번째 요소까지  
print(30 in a)  # 30이 a에 있는지 확인
print(100 in a)  # 100이 a에 있는지 확인
print(100 not in a)  # 100이 a에 없는지 확인

'p' in "python"  # 'p'가 "python"에 있는지 확인
```

```python
# 시퀀스 객체 연결하기

a = [1, 2, 3]
b = [4, 5, 6]
c = a + b  # 리스트 연결
print(c)

d = (1, 2, 3) + (4, 5, 6)  # 튜플 연결
print(d)

e = "Hello" + " " + "World"  # 문자열 연결
print(e)
```

```python
list(range(0, 10, 2)) + list(range(10, 20, 3))  # range 객체 연결
```

```python
# 시퀀스 객체 반복하기

[1, 2, 3] * 3  # 리스트 반복
(1, 2, 3) * 3  # 튜플 반복
"Hello" * 3  # 문자열 반복  
```

```python
# 시퀀스 객체의 요소 갯수 구하기
a = [10, 20, 30, 40, 50, 60, 70, 80, 90]
print(len(a))  # 요소 갯수 구하기
```

```python
print(len(range(0, 10, 2)))  # range 객체의 요소 갯수 구하기
print(len("Hello World"))  # 문자열의 요소 갯수 구하기
print(len((1, 2, 3, 4, 5)))  # 튜플의 요소 갯수 구하기
print(len([1, 2, 3, 4, 5]))  # 리스트의 요소 갯수 구하기
print(len({"a": 1, "b": 2, "c": 3}))  # 딕셔너리의 요소 갯수 구하기
print(len({1, 2, 3, 4, 5}))  # 집합의 요소 갯수 구하기
# print(len(True))  # 불리언의 요소 갯수 구하기
# print(len(None))  # None의 요소 갯수 구하기
# print(len(3.14))  # 실수의 요소 갯수 구하기
# print(len(42))  # 정수의 요소 갯수 구하기
print(len(b"Hello World"))  # 바이트 문자열의 요소 갯수 구하기
print(len(bytearray(b"Hello World")))  # 바이트 배열의 요소 갯수 구하기
```

```python
# 인덱스
a = [38, 21, 53, 62, 19, 85, 47]
print(a[3], a[-3])  # 4번째 요소와 뒤에서 3번째 요소 출력
b = range(0, 10, 2)
print(b[3], b[-3])  # 4번째 요소와 뒤에서 3번째 요소 출력
c="Hello World!"
print(c[11], c[-1])  # 마지막 요소 출력
```

```python
# 요소에 값 할당하기

a = [0, 0, 0, 0]  # 리스트 생성
a[0] = 10  # 첫 번째 요소에 10 할당
a[3] = 40  # 네 번째 요소에 40 할당
print(a)  # [10, 0, 0, 40] 출력

# tuple은 요소에 값을 할당할 수 없으므로 아래 코드는 오류 발생
# b = (0, 0, 0, 0)  # 튜플 생성
# b[0] = 10  # 첫 번째 요소에 10 할당   
```

```python
#del 키워드를 이용하여 요소 삭제하기
a = [10, 20, 30, 40]
del a[0]  # 첫 번째 요소 삭제
del a[-1]  # 마지막 요소 삭제  
# del a[1:3]  # 두 번째 요소부터 세 번째 요소까지 삭제
print(a)  # 오류 발생
```

```python
# slice를 이용하여 요소 삭제하기
a = [10, 20, 30, 40]
a[1:3] = []  # 두 번째 요소부터 세 번째 요소까지 삭제
print(a)  # [10, 40] 출력
```

```python
# 인덱스의 증가폭 사용
a = [10, 20, 30, 40, 50, 60, 70, 80, 90]
print(a[1:7:2])  # [20, 40, 60] 출력
print(a[7:1:-2])  # [80, 60, 40] 출력
```

```python
# 인덱스 생략

a = [10, 20, 30, 40, 50, 60, 70, 80, 90]
print(a[:7])  # 처음부터 7번째 요소까지 출력
print(a[1:7])  # 두 번째 요소부터 일곱 번째 요소까지 출력
print(a[::2])
```
