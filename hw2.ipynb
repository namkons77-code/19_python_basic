```python
col, row = map(int, input().split())
matrix = []
for i in range(row):
    matrix.append(list(input()))
for i in range(row):
    for j in range(col):
        if matrix[i][j] == '*':
            continue
        else:
            matrix[i][j] = 0 # 0을 넣은 뒤에는 요소 주변 8개를 탐색하면서 *이면 요소를 1개씩 증가
            # range에서 생성한 숫자의 마지막 값은 끝나는 값보다 1 작은데
            # 탐색을 위해서 한 칸 아래까지 더 반복해야 하므로 i + 1이 아닌 i + 2가 되어야 함
            for y in range(i - 1, i + 2): # 한 칸 위부터 한 칸 아래까지 반복
                for x in range(j -1, j + 2): # 한 칸 앞(왼쪽)부터 한 칸 뒤(오른쪽)까지 반복
                    if y < 0 or x < 0 or y >= row or x >= col: # 리스트의 범위를 벗어나면 건너뛰기
                        continue
                    elif matrix[y][x] == '*':
                        matrix[i][j] += 1

for i in range(row):
    for j in range(col):
        print(matrix[i][j], end='')
    print()
```

```python
# 문제 14. 성적 관리 프로그램
# 학생들의 성적을 관리하는 프로그램을 작성하시오.

# 다음 학생들의 성적 정보를 딕셔너리로 저장하시오
# 학생1: 이름=김민수, 국어=85, 영어=90, 수학=78
# 학생2: 이름=이영희, 국어=92, 영어=88, 수학=95
# 학생3: 이름=박철수, 국어=78, 영어=85, 수학=82

l = {
    "김민수":{"국어" : 85, "영어" : 90, "수학" :78 },
    "이영희":{"국어" : 92, "영어" : 88, "수학" :95 },
    "박철수":{"국어" : 78, "영어" : 85, "수학" :82 }
}

# 요구사항:
# 1. 각 학생의 총점과 평균을 계산하여 딕셔너리에 추가
for n, s in l.items():
    t = sum(s.values())
    a = t / len(s)
    l[n]["총점"] = t
    l[n]["평균"] = a
print(l)


# 2. 각 학생의 정보를 보기 좋게 출력

print("=== 학생 성적 관리 ===")
for name, info in l.items():
    print(f"이름: {name}")
    print(f"국어: {info['국어']}점, 영어: {info['영어']}점, 수학: {info['수학']}점")
    print(f"총점: {info['총점']}점, 평균: {info['평균']:.2f}점")
    print("---")

# **출력 예시:**

# === 학생 성적 관리 ===
# 이름: 김민수
# 국어: 85점, 영어: 90점, 수학: 78점
# 총점: 253점, 평균: 84.33점
# ---
# 이름: 이영희
# 국어: 92점, 영어: 88점, 수학: 95점
# 총점: 275점, 평균: 91.67점
# ---
# 이름: 박철수
# 국어: 78점, 영어: 85점, 수학: 82점
# 총점: 245점, 평균: 81.67점
# ---


```

```python
# 문제 15. 단어 카운터
# 문자열에서 각 단어가 몇 번 등장하는지 세는 프로그램을 작성하시오.

text = "python is great python is fun python is powerful"

# 요구사항:
# 1. 문자열을 공백으로 분리하여 단어 리스트 만들기
a = text.split()
print(a)

# 2. 각 단어의 등장 횟수 세기
c = {}
for i in a:
    c[i] = c.get(i, 0) + 1
print(c)

# 3. 결과를 출력
for j, c in c.items():
    print(f"{j}: {c}번")

# **출력 예시:**

# python: 3번
# is: 3번
# great: 1번
# fun: 1번
# powerful: 1번
```
