```python
# 
# dict()
zip(["h", "m"],[490, 334])
```

```python
keys = ['a', 'b', 'c']
values = [1, 2, 3]

result = dict(zip(keys, values))
print(result)
# 출력: {'a': 1, 'b': 2, 'c': 3}
```
