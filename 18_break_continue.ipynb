```python
# whie 에서 break 로 반복문 끝내기

i = 0
while True:
    print(i)
    i += 1
    if i == 10:
        break
```

```python
for i in range(10):
    print(i)
    i += 1
    if i == 10:
        break
```

```python
#  continue 로 실행 건너뛰기
for i in range(10):
    if i % 2 == 0:
        continue
    print(i)
```

```python
# 입력한 횟수대로 반복하기
x = int(input())
i = 0
while True:
    print(i)
    i += 1
    if i == x:
        break
```

```python
# 18.6 심사문제
start, stop = map(int, input().split())
i = start
while True:
    if i % 10 == 3:
        i += 1
        continue
    if i > stop:
        break
    print(i, end=" ")
    i += 1


```
