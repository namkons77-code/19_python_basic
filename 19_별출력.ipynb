```python
# 중첩루프 사용하기

for i in range(5):
    for j in range(5):
        print(j,"j", end=" ")
    print(i,"\\\\\\\\\n")
```

```python
for i in range(5):
    for j in range(5):
        if j <= i:
            print("*", end=" ")
    print()
```

```python
for i in range(5):
    for j in range(5):
        if j == i:
            print("*", end=" ")
        else :
            print(" ", end=" ")
    print(i,j)
```

```python
# 19.6 심사문제
# x = int(input())
# y = (x*2)-1

x=3
y=5

for i in range(x):
    for j in range(y):
        if x > j :
            print("_"*(x-i-1), "*" * (2*i + 1))

    print()
```

```python
# 19.6 심사문제
# x = int(input())
# y = (x*2)-1

x=3
y=5

for i in range(x):
    for j in range(y):
        if x == j or x == 1:
            print("*", end=" ")
        else :
            print(" ", end=" ")
    print()
```
