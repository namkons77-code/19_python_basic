```python
# 14.7 심사문제
k, e, m, s = map(int, input().split())

if 0 <= k <=100 and 0 <= e <=100 and 0 <= m <=100 and 0 <= s <=100 : 

    if (k+e+m+s)/4 >= 80 :
        print('합격')
    else :
        print('불합격')
else :
    print('잘못된 점수')
```
